## 1. Sample Data (Replace with your own if desired)

```python
SENTENCES_ZH = [
    "今天是個好天氣，我們一起去台北市立美術館吧！",
    "自然語言處理在教育、醫療與金融領域都有廣泛應用。",
    "請問下週三下午兩點在BS440教室還有課嗎？",
    "你覺得大型語言模型在中文任務上的表現如何？",
    "他知道這件事不要緊",
    "這份報告我寫不好",
    "他有一個女兒，在醫院工作",
    "我在樓上看到了他",
    "兒的生活好痛苦也沒有糧食多病少掙了很多錢"
]
```

## 2. Sentence Segmentation (句子切分)

```python
import re

def split_sentences(text: str):
    """Split Chinese text into sentences using punctuation heuristics."""
    parts = re.split(r"(?<=[。！？!?])", text)
    return [p.strip() for p in parts if p.strip()]

# TODO: Test your function on a paragraph of your own
for s in SENTENCES_ZH:
    print(split_sentences(s))
```

## 3. Tokenization (斷詞)

```python
import jieba

def tokenize_zh(text: str):
    return [tok for tok in jieba.lcut(text) if tok.strip()]

for s in SENTENCES_ZH:
    print(tokenize_zh(s))
```

## 4. POS Tagging (詞性標註)

```python
from jieba import posseg as pseg

def pos_tag_zh(text: str):
    return [(w.word, w.flag) for w in pseg.cut(text)]

for s in SENTENCES_ZH:
    print(pos_tag_zh(s))
```

## 5. Analysis: Frequency & POS Distribution

```python
from collections import Counter

all_tokens = []
all_pos = []
for s in SENTENCES_ZH:
    toks = tokenize_zh(s)
    all_tokens.extend(toks)
    all_pos.extend([p for _, p in pos_tag_zh(s)])

print('Top tokens:', Counter(all_tokens).most_common(10))
print('POS distribution:', Counter(all_pos).most_common())
```

## 6. 比較 jieba 與 spaCy 的斷詞結果

我先用上面的 sample code 跑完 jieba 的斷詞、POS tagging 和詞頻統計，確認都能正常執行後，接著用 spaCy 的 `zh_core_web_sm` 中文模型對同一批句子做比較，看看兩套工具在實際結果上有什麼差異。

### 6.1 用 spaCy 跑同一批句子

```python
import spacy

nlp = spacy.load("zh_core_web_sm")

def tokenize_spacy(text: str):
    doc = nlp(text)
    return [tok.text for tok in doc if tok.text.strip()]

def pos_tag_spacy(text: str):
    doc = nlp(text)
    return [(tok.text, tok.pos_) for tok in doc if tok.text.strip()]

print("=== spaCy Tokenization ===")
for s in SENTENCES_ZH:
    print(tokenize_spacy(s))

print("\n=== spaCy POS Tagging ===")
for s in SENTENCES_ZH:
    print(pos_tag_spacy(s))
```

### 6.2 逐句比較斷詞結果

接下來把兩套工具的斷詞結果放在一起對照，比較容易看出差異在哪裡。

```python
print("=" * 80)
print("jieba vs spaCy — 斷詞比較")
print("=" * 80)

for i, s in enumerate(SENTENCES_ZH):
    jieba_toks = tokenize_zh(s)
    spacy_toks = tokenize_spacy(s)
    print(f"\n--- 句子 {i+1}: {s}")
    print(f"  jieba ({len(jieba_toks)} tokens): {' / '.join(jieba_toks)}")
    print(f"  spaCy ({len(spacy_toks)} tokens): {' / '.join(spacy_toks)}")
    
    # 找出差異
    if jieba_toks != spacy_toks:
        jieba_set = set(jieba_toks)
        spacy_set = set(spacy_toks)
        only_jieba = jieba_set - spacy_set
        only_spacy = spacy_set - jieba_set
        if only_jieba:
            print(f"  → jieba 獨有: {only_jieba}")
        if only_spacy:
            print(f"  → spaCy 獨有: {only_spacy}")
```

### 6.3 詞性標註比較

兩套工具用的 POS 標籤系統完全不同，jieba 用北大標註集（像 `n`, `v`, `r`），spaCy 用 Universal POS Tags（像 `NOUN`, `VERB`, `PRON`），所以不能直接比標籤名稱，但可以看各自標得合不合理。

```python
print("=" * 60)
print("jieba vs spaCy — 詞性標註比較")
print("=" * 60)

for i, s in enumerate(SENTENCES_ZH):
    jieba_pos = pos_tag_zh(s)
    spacy_pos = pos_tag_spacy(s)
    print(f"\n--- 句子 {i+1}: {s}")
    print(f"  jieba : {jieba_pos}")
    print(f"  spaCy : {spacy_pos}")
```

### 6.4 觀察與分析

跑完之後我覺得蠻意外的，一開始以為 spaCy 用了深度學習模型應該會比 jieba 穩，但實際上在這組中文句子裡，spaCy 的斷詞反而切得比較碎、比較不自然。以下挑幾個差異比較明顯的例子來討論，有些句子兩套工具結果差不多就不展開了。

---

**例子 1：「今天是個好天氣，我們一起去台北市立美術館吧！」**

- jieba：`台北市立 / 美術館` → 雖然沒有完整保留「台北市立美術館」，但至少還看得出是一個地名
- spaCy：`台北市 / 立美 / 術館` → 「立美」「術館」完全不是詞，切得很離譜
- 另外 spaCy 還把「個好」黏在一起，把「我們」拆成「我 / 們」，這些都不太合理

在專有名詞的處理上，jieba 明顯比 spaCy 好很多。

---

**例子 2：「請問下週三下午兩點在BS440教室還有課嗎？」**

- jieba：把「BS440」當成一個詞，但出現了「問下」這種不自然的組合
- spaCy：把「BS440」拆成「BS / 440」，然後「課嗎」黏在一起
- 兩套工具都沒辦法完美處理這種中英數混合的句子，各有各的問題

---

**例子 3：「這份報告我寫不好」**

- jieba：`寫 / 不好` → 把「不好」留在一起，語意比較完整
- spaCy：`寫 / 不 / 好` → 拆成三個，失去了「寫不好」作為補語結構的語意

這個差異蠻有趣的，顯示不同工具對中文否定結構、補語的處理邏輯不太一樣。

---

**例子 4：「他知道這件事不要緊」**

- jieba 和 spaCy 都把「不要緊」切成「不要 / 緊」
- 但「不要緊」其實是固定用法（意思是「沒關係」），理想上應該保留為一個詞
- 這代表純詞典式或預設模型的斷詞，碰到慣用語時不一定切得對

---

另外值得注意的是，這批 sample data 裡有幾句本身就帶有歧義，像是「他知道這件事不要緊」「這份報告我寫不好」「我在樓上看到了他」「兒的生活好痛苦也沒有糧食多病少掙了很多錢」。這些句子因為斷詞位置不同，可能會產生不同的語意理解。例如最後一句完全沒有標點，讓兩套工具都很難處理，「病少」這種切法就明顯不對。

---

**整體來看：** 在這次測試中，jieba 的斷詞結果比 spaCy 更接近日常中文的使用習慣，spaCy 雖然用了深度學習模型，但 `zh_core_web_sm` 是比較小的模型，在繁體中文上的表現不太理想。不過兩套工具碰到專有名詞、口語固定搭配、歧義句或中英混合文字時，都還是需要人工確認結果。
